Dataset scope: 1,095 IPL matches and 260,920 deliveries from a public IPL 2008–2024 dataset. The local CSVs in `Dataset/` are cleaned analytical tables used to reproduce the required charts and insights.

## 1. Data Understanding & Cleaning
- Season labels normalized to calendar years.
- Historical franchise names normalized where applicable (e.g., Delhi Daredevils → Delhi Capitals; Kings XI Punjab → Punjab Kings; Royal Challengers Bangalore → Royal Challengers Bengaluru).
- Venue names standardized to reduce duplicate spellings.
- Missing match-result records are excluded from toss-win percentage calculations.
- The analysis covers 17 seasons, 1,095 matches, 37 normalized venues, 260,920 deliveries, 347,756 total runs and 12,950 wicket events in the source analysis.

## Key Business Insights
1. Mumbai Indians lead the source analysis in match wins with 144, followed by Chennai Super Kings (138) and Kolkata Knight Riders (131).
2. 2024 recorded the highest total runs (25,971) and the highest average runs per match (365.79).
3. 2009 had the lowest average runs per match (286.89) and the lowest run rate (7.21).
4. Virat Kohli is the leading run scorer in the 2008–2024 delivery dataset with 8,014 runs.
5. Yuzvendra Chahal is the leading wicket taker in the analysis with 213 wicket events.
6. AB de Villiers has the most Player of the Match awards (25).
7. Teams elected to field after winning the toss in 64.3% of matches; field-first toss winners had a 53.9% match-win rate versus 45.4% when choosing to bat.
8. Total scoring intensity increased substantially: the run rate rose from 7.21 in 2009 to 9.11 in 2024.
9. Wankhede Stadium is the strongest venue in the Mumbai Indians venue-win analysis, with 42 wins.
10. IPL 2022–2024 produced the highest season run totals, reflecting both expanded 74-match seasons and higher scoring rates.
